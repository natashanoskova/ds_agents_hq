# Выбор локации для нефтяной скважины

**Описание проекта**:

Добывающей компании «ГлавРосГосНефть» нужно решить, где бурить новую скважину. 

***Шаги для выбора локации обычно такие:***

- В избранном регионе собирают характеристики для скважин: качество нефти и объём её запасов;

- Строят модель для предсказания объёма запасов в новых скважинах;

- Выбирают скважины с самыми высокими оценками значений;

- Определяют регион с максимальной суммарной прибылью отобранных скважин.

Нам предоставлены пробы нефти в трёх регионах. Характеристики для каждой скважины в регионе уже известны. 

Задача: построить модель для определения региона, где добыча принесёт наибольшую прибыль. Проанализировать возможную прибыль и риски техникой **Bootstrap**.

**Условия проекта**:


Для обучения модели подходит только **линейная регрессия** (остальные — недостаточно предсказуемые).

При разведке региона **исследуют 500 точек**, из которых с помощью машинного обучения **выбирают 200 лучших** для разработки.

Бюджет на разработку скважин в регионе — **10 млрд рублей**.

При нынешних ценах один баррель сырья приносит 450 рублей дохода. **Доход с каждой единицы продукта** составляет **450 тыс. рублей**, поскольку **объём указан в тысячах баррелей**.

После оценки рисков нужно оставить лишь те регионы, в которых **вероятность убытков меньше 2.5%**. Среди них выбрать регион с наибольшей средней прибылью.

*Данные синтетические: детали контрактов и характеристики месторождений не разглашаются.

Для определения нижней и верхней границы выбросов и их удаления используем **IQR**.

```python
def drop_outliers(data, feature):
    
    q25, q75 = data[feature].quantile([0.25, 0.75]).values
    lower = q25 - 1.5 * (q75 - q25)
    upper = q75 + 1.5 * (q75 - q25)
    
    data = data[~(data[feature] < lower)]
    data = data[~(data[feature] > upper)] 
    
    return data
```

## Линейная регрессия, предсказания по регионам

### Регион #0

```python
state = np.random.RandomState(0)
```

```python
def predict_product(data):
    
    target = data['product']
    features = data[['f0', 'f1','f2']]
    
    features_train, features_valid, target_train, target_valid = train_test_split(
    features, target, test_size=0.25, random_state=state)
    
    scaler = StandardScaler()
    scaler.fit(features_train)
    
    features_train = scaler.transform(features_train)
    features_valid = scaler.transform(features_valid)
    
    model = LinearRegression()
    model.fit(features_train, target_train)
    predicted_valid = model.predict(features_valid)
    rmse = mean_squared_error(target_valid, predicted_valid)**0.5
 
    print(f'Размер обучающей выборки          : {features_train.shape}')
    print(f'Размер целевого признака          : {target_train.shape}')
    print()
    print(f'Размер валидационной выборки      : {features_valid.shape}')
    print(f'Размер целевого признака          : {target_valid.shape}')
    print()
    print('Средний запас предсказанного сырья :', predicted_valid.mean())
    print('RMSE модели                        :', rmse)
    
    return (features_train, features_valid, 
            target_train, target_valid,
            predicted_valid, rmse)
```

```python
(features_train_0, features_valid_0,
 target_train_0, target_valid_0,
 predicted_valid_0, rmse_0) = predict_product(data_0)
```

## Подготовка к расчёту прибыли

```python
#Бюджет на разработку скважин в регионе — 10 млрд. рублей

BUDGET_SIZE = 10_000_000_000
```

```python
#Доход с каждой единицы продукта составляет 450 тыс. рублей, т.к. объём указан в тысячах баррелей

PRODUCT_UNIT_INCOME = 450_000
```

```python
#При разведке региона с помощью машинного обучения выбирают 200 лучших точек для разработки

OIL_WELLS_NUMBER = 200
```

```python
#Достаточный объём сырья для безубыточной разработки новой скважины

BUDGET_SIZE / (OIL_WELLS_NUMBER * PRODUCT_UNIT_INCOME)
```

Запас сырья в одной скважине должен быть не менее 111,11 тыс.баррелей.

Средние фактические запасы в каждом регионе:

Регион #0 :   92.833966


Регион #1 :   69.317766


Регион #2 :   94.607042	

## Функция для расчёта прибыли

```python
#Выберите скважины с максимальными значениями предсказаний, финальное количество - 200 скважин
#Просуммируйте целевое значение объёма сырья, соответствующее этим предсказаниям
#Рассчитайте прибыль для полученного объёма сырья

def revenue(target_valid, predicted_valid):
    
    predicted_valid_selected = pd.Series(predicted_valid).sort_values(ascending=False)[:200]
    target_valid_selected = target_valid.reset_index(drop=True)[predicted_valid_selected.index]
    
    sum_product = target_valid_selected.sum()
    product_revenue = sum_product * PRODUCT_UNIT_INCOME - BUDGET_SIZE
    
    return product_revenue
```

## Риски и прибыль для каждого региона

```python
def revenue_distribution(target_valid, predicted_valid):
    
    values = []
    
    for i in range(1000):                                            #Первоначальное количество скважин для отбора n=500
                                                            
        predicted_valid_subsample = pd.Series(predicted_valid).sample(n=500, replace=True, random_state=state) 
        product_revenue = revenue(target_valid, predicted_valid_subsample)                               
        values.append(product_revenue) 
        
    values = pd.Series(values)
    mean_values = values.mean()
    
    lower = values.quantile(0.025) 
    upper = values.quantile(0.975)
    
    losses_risk = values.apply(lambda x: x < 0).sum() / len(values)          #Риски считаем как долю количества 
                                                                             #убыточных позиций от общего количества
    print('Средняя прибыль от запаса сырья :', mean_values.round(2))
    print('Риски убытков                   :', '{:.1%}'.format(losses_risk))
    
    return values, mean_values, lower, upper, losses_risk
```

### Регион #0

```python
values_0, mean_revenue_0, lower_0, upper_0, losses_risk_0 = revenue_distribution(target_valid_0, predicted_valid_0)
```

### Регион #1

```python
values_1, mean_revenue_1, lower_1, upper_1, losses_risk_1 = revenue_distribution(target_valid_1, predicted_valid_1)
```

### Регион #2

```python
values_2, mean_revenue_2, lower_2, upper_2, losses_risk_2 = revenue_distribution(target_valid_2, predicted_valid_2)
```
