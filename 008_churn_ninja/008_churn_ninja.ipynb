# Прогнозирование оттока клиентов телеком-компании 

### Описание проекта

Оператор связи **«Ниединогоразрыва.ком»** хочет научиться прогнозировать отток клиентов. 

Если выяснится, что пользователь планирует уйти, ему будут предложены промокоды и специальные условия. 

Команда оператора собрала персональные данные о некоторых клиентах, информацию об их тарифах и договорах.

**Задача:**

Классификиция клиентов на тех, кто уходит и кто остается. Удержание клиентов и минимизация убытков.

#### Описание услуг

Оператор предоставляет два основных типа услуг: 

1. **Стационарную телефонную связь.** Возможно подключение телефонного аппарата к нескольким линиям одновременно.
2. **Интернет.** Подключение может быть двух типов: через телефонную линию (DSL*,* от англ. *digital subscriber line*, «цифровая абонентская линия») или оптоволоконный кабель (*Fiber optic*).

Также доступны такие услуги:

- **Интернет-безопасность:** антивирус (*DeviceProtection*) и блокировка небезопасных сайтов (*OnlineSecurity*);
- **Выделенная линия технической поддержки** (*TechSupport*);
- **Облачное хранилище файлов для резервного копирования данных** (*OnlineBackup*);
- **Стриминговое телевидение** (*StreamingTV*) и **каталог фильмов** (*StreamingMovies*).

За услуги клиенты могут платить каждый месяц или заключить договор на 1–2 года. Доступны различные способы расчёта и возможность получения электронного чека.

#### Описание данных

Данные состоят из файлов, полученных из разных источников:

- `contract_new.csv` — информация о договоре;
- `personal_new.csv` — персональные данные клиента;
- `internet_new.csv` — информация об интернет-услугах;
- `phone_new.csv` — информация об услугах телефонии.

Во всех файлах столбец `customerID` содержит код клиента.

Информация о договорах актуальна **на 1 февраля 2020**.

**Обозначения признаков:**

- `BeginDate` – дата начала пользования услугами,

- `EndDate` – дата окончания пользования услугами,

- `Type` – тип оплаты: ежемесячный, годовой и тд,

- `PaperlessBilling` – безналичный расчет,

- `PaymentMethod` – способ оплаты,

- `MonthlyCharges` – ежемесячные траты на услуги,

- `TotalCharges` – всего потрачено денег на услуги,

- `Dependents` – наличие иждивенцев,

- `SeniorCitizen` – наличие пенсионного статуса по возрасту,

- `Partner` – наличие супруга(и),

- `MultipleLines` – наличие возможности ведения параллельных линий во время 
звонка

Пропуски в колонках после объединения таблиц связаны с отсутствием подключения пользователя к той или иной услуге, поэтому их можно заполнить значением 'No'.

```python
full = full.fillna('No')
```

Так как даты начала и окончания контракта имею тип данных `'object'`, и, по крайней мере, в колонке даты окончания контракта помимо строковых значений даты есть строковые значения, означающие, что на дату выгрузки данных контракт действителен, воспользуемся параметром `errors ='coerce'` метода `pandas to_datetime` для преобразования типа данных.  

```python
for column in ['BeginDate', 'EndDate']:
    full[column] = pd.to_datetime(full[column], errors ='coerce')
    print(f'Количество пропусков в колонке "{column}" :', full[column].isna().sum())
    print('Тип данных признака :', full[column].dtype)
    print()
```

```python
contract.query('EndDate == "2020-02-01"').customerID.count()
```

В исходных данных нет клиентов, для которых даты выгрузки и завершения контракта совпадают. Пропуски в датах окончания контракта заменим датой выгрузки данных.

```python
full['EndDate'] = full['EndDate'].fillna('2020-02-01').astype('datetime64[ns]')
```

```python
full[['BeginDate', 'EndDate']].describe()
```

Тип данных `'object'` колонки `'TotalCharges'` также вероятнее всего связан с наличием строковых значений.

```python
full['TotalCharges'] = pd.to_numeric(full['TotalCharges'], errors='coerce')

print('Количество пропусков в колонке :', full['TotalCharges'].isna().sum())
```

```python
full.query('TotalCharges.isna()').T
```

Можно заметить, что данные в колонке общих расходов отсутствуют у клиентов, заключивших контракт на услуги в день выгрузки данных. В остальных колонках у этих клиентов пропусков нет, поэтому пропуски можно заполнить соответствующими значениями месячных расходов.

```python
full.loc[full['TotalCharges'].isna(), 'TotalCharges'] = full.loc[full['TotalCharges'].isna(), 'MonthlyCharges']
```

Так как даты начала и окончания договора не могут быть использованы для обучения моделей из-за утечки целевого признака, создадим новый признак - продолжительность клиентского стажа в днях, а так же целевой признак: клиент ушел `(Класс=1)` или нет `(Класс=0)`.

```python
full['Tenure'] = (full['EndDate'] - full['BeginDate']).dt.days
```

```python
full['Left'] = np.where(full['EndDate'] == '2020-02-01', 0, 1)
```

### Подготовка данных для машинного обучения

Выделим признаки и целевой признак в данных.

```python
X = full.drop(['customerID', 'BeginDate', 'EndDate', 'gender','Left'], axis=1) 
y = full['Left']

print('Размер признаков         :', X.shape)
print('Размер целевого признака :', y.shape)
```

```python
categorical = list(X.select_dtypes(include='O').columns)
```

```python
def split(X, y):
    
    features_train, features_test, target_train, target_test = train_test_split(
                                                                       X, y, 
                                                                       test_size=0.25, 
                                                                       stratify=y,
                                                                       random_state=RANDOM_STATE)
    
    print(f'Размер обучающей выборки          : {features_train.shape}')
    print(f'Размер целевого признака          : {target_train.shape}')
    print()
    print(f'Размер тестовой выборки           : {features_test.shape}')
    print(f'Размер целевого признака          : {target_test.shape}')
        
    return features_train, features_test, target_train, target_test
```

**Выводы по разделу:**

- разделены целевой и остальные признаки.

- подготовлена функция сплитования данных на обучающую и тестовую выборку в соотношении 3 : 1. 

### Машинное обучение

Рассмотрим алгоритмы градиентного бустинга **LightGBM** и **CatBoost**, а также более простой алгоритм **RandomForestClassifier.**

Для алгоритма случайного леса используем технику порядкого кодирования категориальных признаков Ordinal Encoding. 

Алгоритмы CatBoost и LightGBM работают с категориальными признаками напрямую.

Масштабирование признаков для алгоритмов, основанных на деревьях не требуется.

#### Градиентный бустинг CatBoost

```python
X_cat = X.copy()

X_cat[categorical] = X[categorical].astype('category')
```

```python
X_train_cat, X_test_cat, y_train_cat, y_test_cat = split(X_cat, y)
```

```python
classes = y_train_cat.unique()

weights = compute_class_weight(class_weight='balanced', classes=classes, y=y_train_cat)
class_weights = dict(zip(classes, weights))
```

```python
grid = {
    'iterations': [1900, 2000], 
    'learning_rate': [0.07, 0.08]
    }
```

```python
%%time

cb_cv = GridSearchCV(estimator = CatBoostClassifier(cat_features=categorical, 
                                                    verbose=False, 
                                                    class_weights=class_weights, 
                                                    random_state=RANDOM_STATE),                      
                     param_grid = grid, 
                     scoring='roc_auc',
                     n_jobs=-1,
                     cv = cv)

cb_cv.fit(X_train_cat, y_train_cat)


print(blk+'Лучшая модель CatBoostClassifier :', ylw+f'{cb_cv.best_params_}'+res)
print(blk+'Лучшее значение метрики AUC-ROC  :', ylw+f'{cb_cv.best_score_}'+res)
print()
```
