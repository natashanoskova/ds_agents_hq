## Подготовка данных

### Обработка пропусков, изменение типов данных

```python
#Приведем названия колонок к нижнему регистру

data.columns = data.columns.str.lower()
```

```python
#Визуализация % пропусков в колонках

missing_values = (data.isna().mean()*100).round(2).sort_values()  

trace = go.Bar(
    x=missing_values.values,
    y=missing_values.index,
    marker_color='crimson',
    orientation = 'h',
    text=missing_values.apply(lambda x: '{}%'.format(x)),           
    textposition='auto',
    hoverinfo='x+y')

layout = {'title': 'Количество пропусков в данных по колонкам датасета, %',
          'xaxis': {'title': 'количество пропусков, %'},
          'yaxis': {'title': 'название колонки'}}
         
fig = go.Figure(data=trace, layout=layout)
fig.show()
```

```python
#Игры, у которых отсутствует оценка критиков, пользователей и рейтинг ESRB по годам

missing_scores_rating = data.query('critic_score.isna() and user_score.isna() and rating.isna()'
                                  ).year_of_release.reset_index()

trace = go.Histogram(
    x=missing_scores_rating.year_of_release,
    marker_color='lightseagreen',
    nbinsx=100)

layout = {'title': 'Игры, у которых отсутствует оценка критиков, пользователей и рейтинг ESRB по годам',
          'xaxis': {'title': 'год'},
          'yaxis': {'title': 'количество пропусков'}}
         
fig = go.Figure(data=trace, layout=layout)
fig.show()
```

```python
#Платформы, у которых отсутствуют данные о дате релиза

missing_year_platforms = data.query('year_of_release.isna()').platform.reset_index()

trace = go.Histogram(
    x=missing_year_platforms.platform,
    marker_color='lightseagreen',
    nbinsx=100)

layout = {'title': 'Платформы, у которых отсутствуют данные о дате релиза',
          'xaxis': {'title': 'название платформы'},
          'yaxis': {'title': 'количество пропусков'}}
         
fig = go.Figure(data=trace, layout=layout)
fig.show()
```

```python
#Жанры, у которых отсутствуют данные о дате релиза

missing_year_genres = data.query('year_of_release.isna()').genre.reset_index()

trace = go.Histogram(
    x=missing_year_genres.genre,
    marker_color='lightseagreen',
    nbinsx=100)

layout = {'title': 'Жанры, у которых отсутствуют данные о дате релиза',
          'xaxis': {'title': 'название жанра'},
          'yaxis': {'title': 'количество пропусков'}}
         
fig = go.Figure(data=trace, layout=layout)
fig.show()
```

Игры, которые не имеют оценок критиков и пользователей, а также возрастной рейтинг есть за каждый год исследуемых данных. Можно предположить, что в любом случае, независимо от возможности оценить игру и присвоить ей рейтинг, каждый год какой-то процент игр остается без оценок, причем, чем больше игр выходит за год, тем больше игр остается без оценок. Кроме того, помимо американского рейтинга ESRB, например в Европе, Японии, Австралии и т.д. есть свои рейтинговые организации.

Данные о дате релиза отсутствуют как для наиболее популярных платформ PS2, DS, PS3, Wii, X360 и жанров Action, Sports, Misc, Role-Playing, Shooter, Adventure, Racing, так и для менее популярных. Также, игры на разных платформах могут иметь разную дату релиза и разные оценки. В названиях некоторых игр, таких как Madden NFL 2002, Madden NFL 2004, PDC World Championship Darts 2008, NASCAR Thunder 2003, FIFA Soccer 2004 и т.д., обозначен год, однако это не означает, что игра на этой платформе вышла в этом году. Заполнение пропусков искусственными значениями, медианой и средним, не очень хорошая идея, и может привести к искажению данных. Поэтому используем заглушки:

оценки критиков не имеют нулевых значений, оценки пользователей имеют одно нулевое значение; для удобства и чтобы не было путаницы, пусть отсутствие оценки будет обозначено как -1,

пропуски в рейтинге как no rating,

пропуски в названии игры и жанре удаляем,

пропуски даты релиза удаляем.

```python
#Обрабатываем пропуски в оценках критиков

data['critic_score'] = data['critic_score'].fillna(-1)
```

```python
#Обрабатываем пропуски в оценках пользователей

data['user_score'] = pd.to_numeric(data['user_score'], errors='coerce').fillna(-1)
```

```python
#Пропуски в рейтинге

data['rating'] = data['rating'].fillna('no rating')
```

```python
#Удаляем пропуски в названии игры (в этих строках пропуск жанра)

data = data[~(data['name'].isna())]
```

## Проверка гипотезы
- Средние пользовательские рейтинги платформ Xbox One и PC одинаковые
- Средние пользовательские рейтинги жанров Action и Sports разные

### Средние пользовательские рейтинги платформ Xbox One и PC одинаковые

Для проверки гипотез будем использовать t-критерий Стьюдента, так как выборки между собой независимы.
Определим значение уровня значимости alpha = 0.05

*Нулевая гипотеза*

H_0: Средний пользовательский рейтинг платформы Xbox One = средний пользовательский рейтинг платформы PC 

*Альтернативная гипотеза*

H_1: Средний пользовательский рейтинг платформы Xbox One ≠ средний пользовательский рейтинг платформы PC

```python
print('Cредний пользовательский рейтинг платформы Xbox One: ', data.query('platform=="XOne" & user_score>-1').user_score.mean())
print('Cредний пользовательский рейтинг платформы PC: ', data.query('platform=="PC" & user_score>-1').user_score.mean())
print()

results = st.ttest_ind(
    data[(data['platform'] == 'XOne') & (data['user_score'] > -1)]['user_score'], 
    data[(data['platform'] == 'PC') & (data['user_score'] > -1)]['user_score'], equal_var = False)   

alpha = 0.05  

print('p-значение:', results.pvalue)

if results.pvalue < alpha:
    print('Отвергаем нулевую гипотезу')
else:
    print('Не получилось отвергнуть нулевую гипотезу') 
```

Средние пользовательские рейтинги не равны, однако вероятность случайно получить такое или большее различие достаточно велика и состаляет примерно 15%, различия между средними не значимы и нулевая гипотеза не отвергается. 

### Средние пользовательские рейтинги жанров Action и Sports разные

Для проверки гипотез в этом случае также будем использовать t-критерий Стьюдента, так как выборки между собой независимы.
Определим значение уровня значимости alpha = 0.05.

Нулевая гипотеза формулируется как равенство независимо от того, как сформулирован вопрос - в виде отрицания или утверждения.

*Нулевая гипотеза*

H_0: Средний пользовательский рейтинг жанра Action = средний пользовательский рейтинг жанра Sports

*Альтернативная гипотеза*

H_1: Средний пользовательский рейтинг жанра Action ≠ средний пользовательский рейтинг жанра Sports

```python
print('Cредний пользовательский рейтинг жанра Action: ', data.query('genre=="Action" & user_score>-1').user_score.mean())
print('Cредний пользовательский рейтинг жанра Sports: ', data.query('genre=="Sports" & user_score>-1').user_score.mean())
print()

results = st.ttest_ind(
    data[(data['genre'] == 'Action') & (data['user_score'] > -1)]['user_score'], 
    data[(data['genre'] == 'Sports') & (data['user_score'] > -1)]['user_score'], equal_var = False)   

alpha = 0.05  

print('p-значение:', results.pvalue)

if results.pvalue < alpha:
    print("Отвергаем нулевую гипотезу")
else:
    print("Не получилось отвергнуть нулевую гипотезу") 
```

**Вывод:**

Средний пользовательский рейтинг платформы Xbox One и средний пользовательский рейтинг платформы PC одинаковы.

Средний пользовательский рейтинг жанра Action и средний пользовательский рейтинг жанра Sports разные.
